```python
import pandas as pd

file = pd.read_csv('C:/Users/hp/OneDrive/Desktop/COS102/archive/AB_NYC_2019.csv')

file.head(5)

file.info()

file.shape
```

```python
to_fill = ['reviews_per_month' , 'last_review']

for col in to_fill:
    if col == 'reviews_per_month':
       file[col] = file[col].fillna(file[col].median())

    elif col == 'last_review':
         file[col] = file[col].fillna(file[col].mode())
file
```

```python
file.isnull().sum()
```

file = file.drop_duplicates()


file.describe()

```python
import matplotlib.pyplot as plt

file['neighbourhood_group'].value_counts().plot(kind='bar')
plt.title('Listings by Neighbourhood Group')
plt.xlabel('Borough')
plt.ylabel('Number of Listings')
plt.show()

file['price'].hist(bins=50, range=(0,1000))

Inference:

Most Airbnb listings in NYC are priced under $500 per night.

There’s a long tail of very expensive listings, but they are rare.

Cutting off at 1000 avoids extreme outliers that would distort the view.


Inference:

Manhattan has the highest number of listings, followed by Brooklyn.

Queens, Bronx, and Staten Island have far fewer listings.

This shows Airbnb activity is concentrated in Manhattan and Brooklyn.
```

```python
import matplotlib.pyplot as plt
file.groupby('neighbourhood_group')['price'].mean().plot(kind='bar')
plt.title('Average Price by Borough')
plt.ylabel('Average Price')
plt.show()

Inference:

Manhattan has the highest average price per night.

Staten Island and Bronx tend to be the cheapest.

Brooklyn and Queens fall in the middle.

This reflects real estate and tourism demand differences across borough
```
