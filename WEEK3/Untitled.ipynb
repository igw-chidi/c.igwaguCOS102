```python
import tkinter as tk
from tkinter import messagebox
import csv

# Load employees from CSV file into a dictionary
def load_employees(filename="GIG-logistics(in).csv"):
    employees = {}
    with open(filename, newline='', encoding="utf-8") as csvfile:
        reader = csv.DictReader(csvfile)
        for row in reader:
            dept = row["Department"].strip()
            name = row["Name"].strip()
            if dept not in employees:
                employees[dept] = []
            employees[dept].append({
                "Name": name,
                "Role": row["Role"].strip(),
                "Email": row["Email"].strip(),
                "Phone": row["Phone"].strip(),
                "Location": row["Location"].strip()
            })
    return employees

employees = load_employees()

def verify_employee():
    name = name_entry.get().strip()
    department = dept_entry.get().strip()

    if department in employees:
        # Find employee in department
        emp = next((e for e in employees[department] if e["Name"].lower() == name.lower()), None)
        if emp:
            # Welcome message with details
            messagebox.showinfo(
                "Welcome",
                f"Welcome {emp['Name']}!\n\n"
                f"Department: {department}\n"
                f"Role: {emp['Role']}\n"
                f"Email: {emp['Email']}\n"
                f"Phone: {emp['Phone']}\n"
                f"Location: {emp['Location']}"
            )

            # Show other members of the department
            others = [e["Name"] for e in employees[department] if e["Name"].lower() != name.lower()]
            if others:
                messagebox.showinfo("Colleagues", f"Other members in {department}: {', '.join(others)}")
            else:
                messagebox.showinfo("Colleagues", f"You are the only member in {department}.")
        else:
            messagebox.showwarning("Not Found", f"{name} does not exist in {department} department.")
    else:
        messagebox.showwarning("Not Found", f"Department '{department}' does not exist.")

# GUI setup
root = tk.Tk()
root.title("Employee Verification System - GIG Logistics")

tk.Label(root, text="Enter Employee Name:").grid(row=0, column=0, padx=10, pady=10)
name_entry = tk.Entry(root)
name_entry.grid(row=0, column=1, padx=10, pady=10)

tk.Label(root, text="Enter Department:").grid(row=1, column=0, padx=10, pady=10)
dept_entry = tk.Entry(root)
dept_entry.grid(row=1, column=1, padx=10, pady=10)

verify_button = tk.Button(root, text="Verify", command=verify_employee)
verify_button.grid(row=2, column=0, columnspan=2, pady=20)

root.mainloop()
```
